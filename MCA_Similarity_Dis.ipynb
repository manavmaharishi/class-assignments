```python
%pip install --upgrade numpy scipy

import numpy as np
from scipy.spatial import distance   # keep for euclidean/manhattan/hamming

# Sample Data
customer_A = [4, 5, 2, 3, 4]
customer_B = [5, 3, 2, 4, 5]

customer_A_binary = np.array([1, 0, 1, 1, 0, 1], dtype=int)
customer_B_binary = np.array([1, 1, 1, 0, 0, 1], dtype=int)

A = np.asarray(customer_A, dtype=float)
B = np.asarray(customer_B, dtype=float)

# 1. Euclidean
euclidean_dist = distance.euclidean(A, B)

# 2. Manhattan
manhattan_dist = distance.cityblock(A, B)

# 3. Cosine (manual)
cos_sim = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

# 4. Hamming (count and proportion)
hamming_prop = distance.hamming(customer_A_binary, customer_B_binary)
hamming_count = hamming_prop * customer_A_binary.size

# 5. Jaccard (manual)
intersection = np.logical_and(customer_A_binary, customer_B_binary).sum()
union = np.logical_or(customer_A_binary, customer_B_binary).sum()
jaccard_sim = intersection / union if union != 0 else 0.0

print("Customer Preference Comparison:\n")
print(f"Euclidean Distance: {euclidean_dist:.2f}")
print(f"Manhattan Distance: {manhattan_dist:.2f}")
print(f"Cosine Similarity: {cos_sim:.4f}")
print(f"Hamming: {hamming_count:.0f} mismatches ({hamming_prop:.4f} proportion)")
print(f"Jaccard Similarity: {jaccard_sim:.4f}")

```
