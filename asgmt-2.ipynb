```python
# Q.4 Understanding Attribute Types Using Python
# Represent different types of attributes with examples

import pandas as pd
import numpy as np

print("=== Q.4: Understanding Attribute Types Using Python ===\n")

# 1. NOMINAL ATTRIBUTES (Categories without order)
print("1. NOMINAL ATTRIBUTES (e.g., names, departments)")

nominal_data = {
    'Employee_ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing']
}

df_nominal = pd.DataFrame(nominal_data)
print("\nNominal Attributes Dataset:")
print(df_nominal)
print(f"Data types:\n{df_nominal.dtypes}\n")

# 2. BINARY ATTRIBUTES (Two possible values)
print("2. BINARY ATTRIBUTES (e.g., gender, pass/fail)")

binary_data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Pass_Fail': ['Pass', 'Fail', 'Pass', 'Pass', 'Fail']
}

df_binary = pd.DataFrame(binary_data)
print("\nBinary Attributes Dataset:")
print(df_binary)
print(f"Data types:\n{df_binary.dtypes}\n")

# 3. ORDINAL ATTRIBUTES (Categories with meaningful order)
print("3. ORDINAL ATTRIBUTES (e.g., low < medium < high)")

ordinal_data = {
    'Employee_ID': [201, 202, 203, 204, 205],
    'Education_Level': ['High School', 'Bachelor', 'Master', 'PhD', 'Bachelor'],
    'Performance_Rating': ['Poor', 'Average', 'Good', 'Excellent', 'Good'],
    'Satisfaction': ['Low', 'Medium', 'High', 'High', 'Medium']
}

df_ordinal = pd.DataFrame(ordinal_data)
print("\nOrdinal Attributes Dataset:")
print(df_ordinal)


# 4. NUMERIC ATTRIBUTES (Continuous and Discrete)
print("4. NUMERIC ATTRIBUTES")

numeric_data = {
    'Product_ID': [301, 302, 303, 304, 305],
    # Continuous attributes
    'Price': [25.99, 45.50, 12.75, 89.99, 33.25],
    'Weight_kg': [2.5, 1.8, 0.9, 3.2, 1.5],
    'Temperature_C': [23.5, 18.7, 31.2, 15.8, 27.3],
    'Height_cm': [165.5, 172.3, 158.9, 180.1, 169.7],
    # Discrete attributes
    'Quantity_Sold': [10, 25, 5, 18, 32],
    'Number_of_Reviews': [15, 42, 8, 67, 23],
    'Stock_Count': [100, 250, 75, 150, 200],
    'Years_in_Service': [2, 5, 1, 8, 3]
}

df_numeric = pd.DataFrame(numeric_data)
print("\nNumeric Attributes Dataset:")
print(df_numeric)
print(f"\nData types:\n{df_numeric.dtypes}")

print("\nContinuous vs Discrete Attributes:")
continuous_attrs = ['Price', 'Weight_kg', 'Temperature_C', 'Height_cm']
discrete_attrs = ['Quantity_Sold', 'Number_of_Reviews', 'Stock_Count', 'Years_in_Service']

print(f"Continuous Attributes: {continuous_attrs}")
print(f"Discrete Attributes: {discrete_attrs}")
```
